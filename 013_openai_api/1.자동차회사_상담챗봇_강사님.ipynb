### 1. 코드 소개 및 개요

이 코드는 **자동차 회사의 정보를 제공하는 챗봇**을 구현한 예제입니다. 챗봇은 자동차 모델, 서비스, 위치, 연락처 등 자동차 회사의 다양한 정보를 바탕으로 사용자의 질문에 답변합니다. 

코드는 크게 다음 세 부분으로 구성됩니다:
1. **데이터베이스**: 자동차 회사 정보를 저장합니다.
2. **함수**: 사용자의 입력을 처리하고 OpenAI API를 통해 답변을 생성합니다.
3. **대화 루프**: 사용자와 지속적으로 대화할 수 있도록 반복 실행됩니다.

### 2. 핵심 코드 분석

#### (1) 자동차 회사 데이터베이스 정의
- `car_company_info` 딕셔너리를 사용하여 자동차 회사 정보를 구조적으로 저장합니다.
- 주요 키와 값은 다음과 같습니다:

| 키       | 내용                                                          | 예제 값                                    |
|----------|---------------------------------------------------------------|------------------------------------------|
| `회사`    | 회사 이름                                                     | "구름자동차"                              |
| `모델`    | 차량 모델과 가격 정보                                         | "아반떼: 1000만원"                        |
| `서비스`  | 정기 점검, 긴급 출동, 보증 기간 등 서비스 제공 내용            | "정기점검: 6개월마다 무료 정기점검"       |
| `연락처`  | 고객센터 및 서비스센터 번호                                   | "1588-1234"                              |
| `위치`    | 회사 위치                                                     | "서울특별시 강남구 테헤란로 123"          |

> **구조적인 데이터베이스**를 사용하여 다양한 정보를 효율적으로 저장하고 활용할 수 있습니다.

#### (2) `get_car_chatbot_response` 함수
- 사용자 입력(`user_input`)을 기반으로 챗봇 응답을 생성하는 역할을 합니다.
- 주요 작업:
  1. **자동차 회사 데이터 JSON 변환**: `car_company_info`를 JSON 형식으로 변환하여 `prompt`에 포함합니다.
  2. **OpenAI API 호출**: GPT 모델에 질문(`user_input`)과 회사 정보를 전달하여 응답을 생성합니다.
  3. **응답 반환**: OpenAI API의 답변을 사용자에게 출력합니다.

> OpenAI API를 통해 **데이터 기반 맞춤형 응답**을 생성합니다.

#### (3) 대화 루프
- 사용자의 입력을 반복적으로 받아 챗봇과 대화할 수 있는 구조를 만듭니다.
- "종료"라는 명령어를 입력하면 대화가 종료됩니다.

```python
while True:
    user_input = input("User: ")
    if user_input == "종료":
        print("대화를 종료합니다. 감사합니다!")
        break
    response = get_car_chatbot_response(user_input)
    print(f"Assistant: {response}")
```

### 3. OpenAI API 사용 방식

- **Prompt 설계**: 챗봇이 `car_company_info` 데이터를 참조하도록 설계되었습니다. 예시:
  ```json
  {
      "role": "system",
      "content": "당신은 자동차 회사의 전문 상담사입니다. ... "
  }
  ```
- **파라미터 설정**:
  - `model`: `gpt-4o-mini` 사용 (모델 이름은 `MODEL` 변수 한 곳에서 관리).
  - `max_completion_tokens`: 응답 길이를 제한. 너무 작으면 답변이 중간에 잘립니다.
  - `temperature`: 응답의 무작위성을 조정. 정해진 정보를 안내하는 상담 챗봇이므로 0.3으로 낮게 설정.

### 5. 실행 예제

**사용자**: 소나타 가격이 얼마인가요?  
**챗봇**: 소나타의 가격은 2000만원입니다.  

**사용자**: 고객센터 번호가 어떻게 되나요?  
**챗봇**: 고객센터 번호는 1588-1234입니다.  

> 이 코드는 자동차 회사 상담 챗봇 구현의 기초가 되며, 다양한 확장 가능성을 제공합니다.

````python
# 최초 실행 시 필수 패키지 설치
# openai: OpenAI API 호출 라이브러리 / python-dotenv: .env 파일의 API 키를 읽는 라이브러리
%pip install -q -U openai python-dotenv
````

````python
# 자동차 회사의 정보를 담은 데이터베이스
# 딕셔너리 안에 딕셔너리를 넣어 정보를 분류별로 정리 (키: 분류 이름, 값: 내용)
car_company_info = {
    "회사": "구름자동차",
    # 차량 모델(차종)과 가격
    "모델": {
        "아반떼(준중형 세단)" : "1000만원",
        "소나타(중형 세단)" : "2000만원",
        "그랜저(준대형 세단)" : "3000만원",
        "팰리세이드(대형 SUV)" : "4000만원"
    },
    # 고객 서비스 안내
    "서비스": {
        "정기점검": "6개월마다 무료 정기점검을 제공합니다.",
        "긴급출동": "24시간 긴급출동 서비스를 제공합니다.",
        "보증기간": "모든 차량에 대해 5년 또는 10만 km 보증을 제공합니다."
    },
    # 연락처 (고객센터 번호는 상담 프롬프트에서도 사용)
    "연락처": {
        "고객센터": "1588-1234",
        "서비스센터": "02-123-4567"
    },
    "위치": "서울특별시 강남구 테헤란로 123"
}
````

````python
from openai import OpenAI, OpenAIError   # API 클라이언트와 API 오류 클래스
from dotenv import load_dotenv            # .env 파일을 읽어 환경 변수로 등록
import json                               # 딕셔너리를 JSON 문자열로 변환
import os                                 # 환경 변수 읽기

# .env 파일에서 API 키 불러오기
from pathlib import Path
# 현재 폴더의 .env를 먼저 찾고, 없으면 상위 폴더의 .env를 사용
load_dotenv(Path.cwd() / '.env') or load_dotenv(Path.cwd().parent / '.env')
api_key = os.getenv("OPENAI_API_KEY", "").strip()
# 키가 없거나 .env.example의 예시 문구를 그대로 둔 경우 실행을 멈춤
if not api_key or api_key.startswith("운영팀에서_"):
    raise ValueError(".env 파일에 OPENAI_API_KEY가 없습니다.")

# API를 호출할 클라이언트 객체 생성
client = OpenAI(api_key=api_key)

# 사용할 모델은 한 곳에서 관리
# gpt-3.5-turbo-0125는 2026-10-23 서비스 종료 예정이라 gpt-4o-mini로 교체
MODEL = "gpt-4o-mini"


def get_car_chatbot_response(user_input):
    # 자동차 회사 정보를 JSON 문자열로 변환해 system 메시지에 포함
    # ensure_ascii=False: 한글을 그대로 유지 / indent=4: 보기 좋게 들여쓰기
    car_info_json = json.dumps(car_company_info, ensure_ascii=False, indent=4)
    # system 메시지: 챗봇의 역할과 답변 규칙, 참고할 회사 정보를 지정
    # 고객센터 번호는 데이터에서 읽어 오므로 데이터만 바꾸면 프롬프트도 함께 바뀜
    system_prompt = (
        f"당신은 자동차 회사의 전문 상담사입니다. 아래 제공된 자동차 회사 정보에 근거하여 성실하고 친절하게 답변해 주세요. 제공된 정보에 없는 가격이나 사양은 임의로 추측하지 말고 고객센터({car_company_info['연락처']['고객센터']}) 문의를 안내하세요.\n"
        "다음은 자동차 회사의 정보입니다.\n"
        f"{car_info_json}"
    )

    # API에 보낼 메시지 목록 (system: 역할 지정, user: 사용자 질문)
    # 매번 새로 만들기 때문에 이전 대화는 기억하지 못함
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_input}
    ]

    try:
        # OpenAI API 호출
        response = client.chat.completions.create(
            model=MODEL,
            messages=messages,
            max_completion_tokens=500,   # 150이면 긴 답변이 중간에 잘림
            temperature=0.3,             # 낮을수록 일관된 답변 (정보 안내용이라 낮게 설정)
            frequency_penalty=0.1,       # 같은 단어 반복을 약간 억제
            presence_penalty=0.1         # 이미 나온 주제 반복을 약간 억제
        )
        # 응답 중 첫 번째 답변의 텍스트만 꺼내서 반환
        return response.choices[0].message.content

    except OpenAIError as e:
        # 오류 시 None 대신 안내 문구를 돌려줌
        return f"API 호출 중 오류가 발생했습니다: {e}"
````

````python
# 메인 대화 루프
print("구름 자동차 상담 챗봇에 오신 것을 환영합니다! 종료를 원하시면 '종료'라고 입력하세요.")
while True:   # "종료"를 입력할 때까지 계속 반복
    # 사용자 입력 받기 (strip: 앞뒤 공백 제거)
    user_input = input("User: ").strip()

    # "종료" 입력 시 대화 종료
    if user_input == "종료":
        print("대화를 종료합니다. 감사합니다!")
        break   # 반복문 탈출

    # 빈 입력은 API를 호출하지 않음
    if not user_input:
        print("질문을 입력해 주세요.")
        continue   # 아래 코드를 건너뛰고 다시 입력 받기

    # 챗봇 응답 생성 후 출력
    response = get_car_chatbot_response(user_input)
    print(f"Assistant: {response}")
````

````python
# 자동차 회사의 정보를 담은 데이터베이스
# (위와 같은 데이터. 이 부분부터 따로 실행해도 동작하도록 다시 정의)
car_company_info = {
    "회사": "구름자동차",
    # 차량 모델(차종)과 가격
    "모델": {
        "아반떼(준중형 세단)" : "1000만원",
        "소나타(중형 세단)" : "2000만원",
        "그랜저(준대형 세단)" : "3000만원",
        "팰리세이드(대형 SUV)" : "4000만원"
    },
    # 고객 서비스 안내
    "서비스": {
        "정기점검": "6개월마다 무료 정기점검을 제공합니다.",
        "긴급출동": "24시간 긴급출동 서비스를 제공합니다.",
        "보증기간": "모든 차량에 대해 5년 또는 10만 km 보증을 제공합니다."
    },
    # 연락처
    "연락처": {
        "고객센터": "1588-1234",
        "서비스센터": "02-123-4567"
    },
    "위치": "서울특별시 강남구 테헤란로 123"
}
````

````python
# 대화 맥락을 유지하기 위해 전체 메시지를 저장하는 리스트
# 회사 정보를 JSON 문자열로 변환
car_info_json = json.dumps(car_company_info, ensure_ascii=False, indent=4)
# 챗봇의 역할과 답변 규칙, 회사 정보를 담은 system 메시지
SYSTEM_PROMPT = (
    f"당신은 자동차 회사의 전문 상담사입니다. 아래 제공된 자동차 회사 정보에 근거하여 성실하고 친절하게 답변해 주세요. 제공된 정보에 없는 가격이나 사양은 임의로 추측하지 말고 고객센터({car_company_info['연락처']['고객센터']}) 문의를 안내하세요.\n"
    "다음은 자동차 회사의 정보입니다.\n"
    f"{car_info_json}"
)

# 대화 기록 리스트: system 메시지로 시작해서 user, assistant 메시지가 차례로 쌓임
# 함수 밖(전역)에 두어야 함수를 여러 번 호출해도 기록이 사라지지 않음
messages = [{"role": "system", "content": SYSTEM_PROMPT}]


def reset_conversation():
    # 대화 기록 초기화 (system 메시지만 남김)
    # messages[1:]: 0번(system)을 제외한 나머지 전부
    del messages[1:]


def get_car_chatbot_response(user_input, car_company_info):
    # 회사 정보가 바뀌었을 수 있으므로 system 메시지를 매번 갱신
    info = json.dumps(car_company_info, ensure_ascii=False, indent=4)
    messages[0]["content"] = (
        f"당신은 자동차 회사의 전문 상담사입니다. 아래 제공된 자동차 회사 정보에 근거하여 성실하고 친절하게 답변해 주세요. 제공된 정보에 없는 가격이나 사양은 임의로 추측하지 말고 고객센터({car_company_info['연락처']['고객센터']}) 문의를 안내하세요.\n"
        "다음은 자동차 회사의 정보입니다.\n"
        f"{info}"
    )

    # 사용자 메시지를 대화 기록에 추가
    messages.append({"role": "user", "content": user_input})

    try:
        # 지금까지의 대화 기록 전체를 함께 보내야 모델이 앞의 대화를 참고할 수 있음
        response = client.chat.completions.create(
            model=MODEL,
            messages=messages,          # 이전 대화 전체를 함께 전달
            max_completion_tokens=500,  # 답변 최대 길이
            temperature=0.3,            # 낮을수록 일관된 답변
            frequency_penalty=0.1,      # 같은 단어 반복 억제
            presence_penalty=0.1        # 같은 주제 반복 억제
        )
    except OpenAIError as e:
        # 실패한 질문은 기록에서 제거해 user 메시지가 연속으로 쌓이지 않게 함
        messages.pop()
        return f"API 호출 중 오류가 발생했습니다: {e}"

    # 모델의 응답도 대화 기록에 추가
    assistant_response = response.choices[0].message.content
    messages.append({"role": "assistant", "content": assistant_response})
    return assistant_response
````

````python
reset_conversation()  # 셀을 다시 실행할 때 이전 대화 초기화

# 메인 대화 루프
print("구름 자동차 상담 챗봇에 오신 것을 환영합니다! 종료를 원하시면 '종료'라고 입력하세요.")
while True:   # "종료"를 입력할 때까지 계속 반복
    # 사용자 입력 받기 (strip: 앞뒤 공백 제거)
    user_input = input("User: ").strip()

    # "종료" 입력 시 대화 종료
    if user_input == "종료":
        print("대화를 종료합니다. 감사합니다!")
        break

    # 빈 입력은 API를 호출하지 않음
    if not user_input:
        print("질문을 입력해 주세요.")
        continue

    # 맥락 유지 버전 함수 호출 (회사 정보도 함께 전달)
    response = get_car_chatbot_response(user_input, car_company_info)
    print(f"Assistant: {response}")
````
