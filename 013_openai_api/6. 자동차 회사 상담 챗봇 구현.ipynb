```python
# 강사님 예시 코드
from openai import OpenAI
from dotenv import load_dotenv
import os

# .env 파일에서 API 키 불러오기
load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")
if not api_key:
    raise ValueError(".env 파일에 OPENAI_API_KEY가 없습니다.")

client = OpenAI(api_key=api_key)


# TXT 파일을 읽어 QnA 형식으로 변환
def load_and_convert_txt(file_path):
    qna_pairs = []

    # 파일을 읽고 질문과 답변을 분리
    with open(file_path, "r", encoding="utf-8") as file:
        for line in file:
            line = line.strip()

            # 빈 줄이나 구분자(:)가 없는 줄은 제외
            if not line or ":" not in line:
                continue

            question, answer = map(str.strip, line.split(":", 1))

            # 질문이나 답변이 비어 있으면 제외
            if not question or not answer:
                continue

            qna_pairs.append({
                "question": question,
                "answer": answer
            })

    return qna_pairs

# 로컬 QnA 검색 및 OpenAI API 답변 생성
def generate_response(user_input, qna_data):

    # 동일한 질문이 있는지 먼저 검색
    for qna in qna_data:
        if user_input == qna["question"]:
            return qna["answer"]

    # QnA 전체를 참고 자료로 변환
    context = "\n".join(
        [f"{q['question']}: {q['answer']}" for q in qna_data]
    )

    # 사용자 질문과 참고 자료로 프롬프트 구성
    prompt = f"""
다음은 ICPCNEW에 대한 정보입니다.

{context}

위의 정보만 참고하여 사용자 질문에 답변해 주세요.
관련 정보가 없다면 "제공된 자료에서 확인할 수 없습니다."라고 답변하세요.

질문: {user_input}
답변:
"""

    # OpenAI API 호출
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {
                "role": "system",
                "content": "당신은 ICPCNEW 대회 관련 전문 챗봇입니다."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        max_completion_tokens=300,
        temperature=0.3
    )

    return response.choices[0].message.content


# 메인 실행
if __name__ == "__main__":

    file_path = "ICPCNEW_QnA_Formatted.txt"

    # 파일 존재 여부 확인
    if not os.path.isfile(file_path):
        raise FileNotFoundError(f"파일을 찾을 수 없습니다: {file_path}")

    # QnA 데이터 불러오기
    qna_data = load_and_convert_txt(file_path)

    if not qna_data:
        raise ValueError("사용 가능한 QnA 데이터가 없습니다.")

    print(f"QnA 데이터 {len(qna_data)}개 로드 완료")
    print("ICPCNEW 챗봇 시작! 종료하려면 '종료'를 입력하세요.")

    while True:
        user_input = input("\nUser: ").strip()

        # 종료 명령 처리
        if user_input == "종료":
            print("대화를 종료합니다.")
            break

        # 빈 질문 제외
        if not user_input:
            print("질문을 입력해 주세요.")
            continue

        # 질문에 대한 답변 생성
        try:
            answer = generate_response(user_input, qna_data)
            print(f"Answer: {answer}\n")
        except Exception as e:
            print(f"API 호출 중 오류가 발생했습니다: {e}")
```
