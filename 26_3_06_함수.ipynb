### 실습_19h40
사용자로부터 점수 세개를 입력 받습니다(국어,영어,수학)
세 점수의 총점과 평균을 리턴하는 함수를 만듭니다.
함수를 호출하여 넘겨 받은 값을 활요하여 아래와 같이 화면에 표시 합니다 세 점수 xx, xx, xx 의 총점은 xxx, 평균은 xx.x 입니다

```python
# 함수 정의
def calc_scores(ss):
    sum = 0    
    for s in ss:
        sum += s

    average = sum / len(ss)

    return sum, average

# 변수 정의
subjects = ["국어", "영어", "수학"]
scores = []

# 변수 값 대입
for subject in subjects:
    score = int(input(f"{subject} 점수를 입력하세요: "))
    scores.append(score)

# 함수 호출
sum, average = calc_scores(scores)
print(average)
print(f"세 점수 {scores}의 총점은 {sum}, 평균은 {average:.1f}입니다")
```
