전체 시간이 `3725`초일 때, 이것을 **시간 / 분 / 초** 로 나누어 아래와 같이 출력하세요.

```
1시간 2분 5초
```

힌트: 1시간은 3600초, 1분은 60초입니다. `//` 와 `%` 를 사용하세요.

---

````python
total_sec = 3725

hour = total_sec // (60*60)
rest = total_sec % (60*60)

minute = rest // 60
second = rest % 60

print(hour, "시간", minute, "분", second, "초")
````

user로 부터 숫자 두개를 입력받아 두수를 덧셈하는 코드를 작성 하시오  
예)  
첫번째 숫자를 입력하세요 : 10  
두번째 숫자를 입력하세요 : 20  
  
두수의 합은 30입니다

````python
num1 = int(input("첫번째 숫자를 입력하세요: "))
num2 = int(input("두번째 숫자를 입력하세요: "))

total = num1 + num2

print("두 수의 합은", total, "입니다.")
print(f"두 수의 합은 {total} 입니다.")
````
