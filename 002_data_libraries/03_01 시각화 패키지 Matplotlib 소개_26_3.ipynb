```python
X = np.linspace(-np.pi, np.pi, 256)
C, S = np.cos(X), np.sin(X)
plt.title("legend를 표시한 plot")
plt.plot(X, C, ls="--", label="cosine")
plt.plot(X, S, ls=":", label="sine")
plt.xlabel("time")
plt.ylabel("amplitude")
plt.legend(loc=0)
plt.show()
```

```python
x1 = np.linspace(0.0, 5.0)
x2 = np.linspace(0.0, 2.0)
y1 = np.cos(2 * np.pi * x1) * np.exp(-x1)
y2 = np.cos(2 * np.pi * x2)

ax1 = plt.subplot(2, 1, 1)  ## 2행 1열 행렬의 첫번째(위쪽에 위치)
plt.plot(x1, y1, 'yo-') # 노란색 동그라미
plt.title('A tale of 2 subplots')
# plt.xlabel('time (s)')
plt.ylabel('Damped oscillation')
print(ax1)

ax2 = plt.subplot(2, 1, 2)   ## 2행 1열 행렬의 두번째(애래쪽에 위치)
plt.plot(x2, y2, 'r.-') # 빨간색 ___
plt.xlabel('time (s)')
plt.ylabel('Undamped')
print(ax2)

plt.tight_layout()   # 여백을 줄여줌
plt.show()
```

```python
import time

# 현재 시간을 정수로 변환하여 시드값으로 설정 -> 매번 실행할 때마다 다른 결과가 나오도록
seed_value = int(time.time()) # 초 단위(vs 밀리초 단위로도 가져올 수 있음)
np.random.seed(seed_value)
print(seed_value) # 1790073118 -> 1790073136..

plt.subplot(221)   ## 2x2 행렬의 좌상단
plt.plot(np.random.rand(5))
plt.title("axes 1")

plt.subplot(222)    ## 2x2 행렬의 우상단
plt.plot(np.random.rand(5))
plt.title("axes 2")

plt.subplot(223)    ## 2x2 행렬의 좌하단
plt.plot(np.random.rand(5))
plt.title("axes 3")

plt.subplot(224)    ## 2x2 행렬의 우하단
plt.plot(np.random.rand(5))
plt.title("axes 4")

plt.tight_layout()
plt.show()
```

```python
fig, ax0 = plt.subplots() # fig = 그림 그리는 도화지, ax0 = 실제 그래프
ax1 = ax0.twinx()  # x축을 공유하는 새로운 Axes

ax0.set_title("2개의 y축 한 figure에서 사용하기")

ax0.plot([10, 5, 2, 9, 7], 'r-', label="y0") 
ax0.set_ylabel("y0")
ax0.grid(True)

ax1.plot([100, 200, 220, 180, 120], 'g:', label="y1")
ax1.set_ylabel("y1")
ax1.grid(True)

ax0.set_xlabel("공유되는 x축")

plt.show()
```
