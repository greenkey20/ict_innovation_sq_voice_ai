## 조건의 재료: 비교 연산자

조건문은 "참인가 거짓인가"를 판단해서 실행할 코드를 고릅니다.
그 판단의 재료가 비교 연산자입니다.

| 연산자 | 의미 |
|---|---|
| `==` | 같다 |
| `!=` | 다르다 |
| `>` | 크다 |
| `<` | 작다 |
| `>=` | 크거나 같다 |
| `<=` | 작거나 같다 |

`=`는 값을 대입하는 기호이고, `==`가 비교하는 기호입니다. 자주 혼동하는 부분입니다.

````python
data = 10

print("시작")
# == 같다
# > 크다
# < 작다
#if data > 20:  # 조건이 거짓
if data > 5:   # 조건이 참
    print("10보다 큽니다") 
    print("확인 완료")

print("종료")
````

## if ~ else

`if`만 쓰면 조건이 거짓일 때 아무 일도 일어나지 않습니다.
거짓일 때 할 일을 정하려면 `else`를 붙입니다.

```
if 조건:
    참일 때 실행
else:
    거짓일 때 실행
```

`else` 뒤에도 콜론이 붙고, 조건은 쓰지 않습니다.

````python
#if else 분기를 사용
data = int(input("숫자를 입력 하세요 : "))
if data > 0:
    print("0보다 큽니다")
else:
    print("0보다 작습니다")
````

## if ~ elif ~ else

경우의 수가 셋 이상일 때 사용합니다.
`elif`는 "else if"의 줄임말이며, 필요한 만큼 여러 번 쓸 수 있습니다.

```
if 조건1:
    조건1이 참일 때
elif 조건2:
    조건1은 거짓, 조건2가 참일 때
elif 조건3:
    ...
else:
    모두 거짓일 때
```
---

````python
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"점수 {score}점의 학점은 {grade}입니다.")
````

## 논리 연산자

조건을 두 개 이상 묶을 때 사용합니다.

| 연산자 | 의미 | 결과가 참인 경우 |
|---|---|---|
| `and` | 그리고 | 양쪽 조건이 모두 참 |
| `or` | 또는 | 한쪽이라도 참 |
| `not` | 부정 | 참이면 거짓, 거짓이면 참 |

````python
data = 10
print(data > 20)
print(data > 0)

print(data > 20 or data > 0)
print(data > 20 and data > 0)
print(not(data > 20))
````
