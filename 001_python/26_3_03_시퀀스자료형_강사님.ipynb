## 1. 시퀀스란

값이 **순서대로** 나열된 자료형을 시퀀스라고 합니다.
문자열, 리스트, 튜플, range가 여기에 속합니다.

종류는 다르지만 다음 기능을 공통으로 사용할 수 있습니다.

| 기능 | 예 |
|---|---|
| 길이 확인 | `len(값)` |
| 인덱싱 | `값[0]` |
| 슬라이싱 | `값[1:3]` |
| 포함 확인 | `"a" in 값` |
| 연결 | `값1 + 값2` |
| 반복 | `값 * 3` |

| 자료형 | 예 | 주요 용도 | 요소 교체 |
|---|---|---|---|
| `list` | `[80, 90, 70]` | 수정하며 쓰는 목록 | 가능 |
| `tuple` | `(10, 20)` | 구성이 고정된 묶음 | 불가 |
| `range` | `range(1, 6)` | 규칙적인 정수 범위 | 불가 |
| `str` | `"Python"` | 문자열 | 불가 |

순서가 있다는 말은 크기순으로 정렬되어 있다는 뜻이 아니라,
**위치 번호로 꺼낼 수 있다**는 뜻입니다. 같은 값이 여러 번 들어 있어도 됩니다.

```python
text = "파이썬"    #문자열
numbers = [10, 20, 30]   #리스트
point = (3, 5)    #튜플

print(len(text), len(numbers), len(point))
print(text[0], numbers[0], point[0])
print("파" in text, 20 in numbers, 40 in numbers)
```

```python
no = [10,20,30,40,50,60]
print(no[0], no[2], no[4])
print(no[-6], no[-4], no[-2])
```

```python
word = "Python"
print(word[0:3])
print(word[2:5])
print(word[:5])
print(word[2:])
print(word[::2])
print(word[-4::2])
```

2. 시퀀스 자료형의 인덱스 2줄 정리
3. 시퀀스 자료형 중 수정이 가능한 자료형은?

1. 시퀀스 자료형 4가지와 특징
* `list` : 여러 데이터를 순서대로 저장하며, 저장된 값을 수정할 수 있습니다.
* `tuple` : 여러 데이터를 순서대로 저장하지만, 저장된 값을 수정할 수 없습니다.
* `range` : 일정한 규칙의 정수 범위를 만들어 반복문 등에 사용합니다.
* `string` : 여러 문자를 순서대로 저장하는 문자열 자료형이며, 문자 자체는 수정할 수 없습니다.

2. 시퀀스 자료형의 인덱스
   인덱스는 시퀀스 자료형의 각 데이터에 붙는 위치 번호이며, 첫 번째 데이터의 인덱스는 `0`부터 시작합니다.
   음수 인덱스를 사용하면 뒤에서부터 접근할 수 있으며, 마지막 데이터는 `-1`입니다.

3. 수정이 가능한 시퀀스 자료형
   **`list`만 수정 가능합니다.**
   `tuple`, `range`, `string`은 생성된 후 값을 변경할 수 없습니다.
