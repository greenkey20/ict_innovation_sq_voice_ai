* python print() 함수는 문자열들을 comma로 나열하면 default로 공백 1개가 들어간다

````python
print("2026","07","25")
print("2026","07","25",sep=" ") # default
print("2026","07","25",sep="")
print("2026","07","25",sep="-")
````

````python
print(2026,7,25)
print(2026,7,25,sep="*")
````

print() 뒤에는 줄바꿈이 항상 존재한다

````python
print("첫 번째")
print("두 번째")
````

````python
print("첫 번째",end="\n") # default
print("두 번째")
````

````python
print("첫 번째",end="")
print("두 번째")
````

````python
print("첫 번째",end="~")
print("두 번째")
````

전체 시간이 `3725`초일 때, 이것을 **시간 / 분 / 초**로 나누어 아래와 같이 출력하세요.

```
1시간 2분 5초
```

힌트: 1시간은 3600초, 1분은 60초입니다. `//`와 `%` 연산자를 사용하세요.

````python
print(3725 // 3600, "시간", (3725 % 3600) // 60, "분", (3725 % 3600) % 60, "초", sep=" ")
````
