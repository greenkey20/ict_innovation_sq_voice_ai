## 함수 만들기: def

가장 단순한 형태부터 시작합니다.

```
def 함수이름():
    실행할 코드
```

- `def`로 시작하고 함수 이름을 붙입니다.
- 이름 뒤에 괄호와 콜론이 옵니다.
- 실행할 코드는 들여씁니다.

함수를 정의하는 것만으로는 아무 일도 일어나지 않습니다.
**이름 뒤에 괄호를 붙여 호출해야** 실행됩니다.

````python
def Greet():   # 함수 정의
    print("안녕하세요!")

Greet()   #함수 호출
````

2. 세 점수의 총점과 평균을 리턴하는 함수를 만듭니다.
3. 함수를 호출하여 넘겨 받은 값을 활요하여 아래와 같이 화면에 표시 합니다
   세 점수 xx, xx, xx 의 총점은 xxx, 평균은 xx.x 입니다

````python
def ave(num1, num2, num3):
  return (num1 + num2 + num3), (num1 + num2 + num3)/3

num1 = int(input("첫 번째 숫자 입력"))
num2 = int(input("두 번째 숫자 입력"))
num3 = int(input("세 번째 숫자 입력"))

print(f"세 점수 {num1}, {num2}, {num3} 의 총점은 {ave(num1, num2, num3)[0]}, 평균은 {ave(num1, num2, num3)[1]:.1f} 입니다.")
````
