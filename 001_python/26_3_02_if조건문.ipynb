2. 입력받은 정수가 홀수인지 짝수인지 판단해서 결과를 화면에 표시합니다.
3. 0은 짝수로 취급합니다.

```python
data = int(input("자연수를 입력하세요 : "))
if data % 2 == 0:
    print("입력한 자연수는 짝수입니다")
else:
    print("입력한 자연수는 홀수입니다")
```

```python
score = 59.9999
# grade = "F"

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else: # 마지막 else는 예외 처리에 많이 사용함
    grade = "F"

print(f"점수 {score}점의 학점은 {grade}입니다")
```

### 중첩된 조건문/if문

```python
is_member = True # name 'true' is not defined
amount = 12000

if is_member:
    print("회원입니다")
    if amount >= 30000:
        print("무료 배송 대상입니다(구매금액 3만원 이상)")
    else:
        print("배송비 3천원이 추가됩니다")
else:
    print("회원이 아닙니다 -> 배송비 4천원이 추가됩니다")
```
