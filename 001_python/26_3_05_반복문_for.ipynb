```python
for i in range(1,10):
    result = 3 * i
    print(f"3 x {i} = {result}")
```

```python
# 반복할 내용이 미정일 때 등..
for item in []:
    print("이 줄은 실행되지 않습니다")

print("반복문 순회 끝")
```

```python
scores = [90, 85, 77, 92, 68]
sum = 0
# 평균 값 담을 변수도 미리 만들어두는 게 좋다

for score in scores:
    sum += score

average = sum / len(scores)

print(f"{scores}의 평균은 {average}입니다")
```
