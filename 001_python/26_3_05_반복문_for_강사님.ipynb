## for 기본 구조

```
for 변수 in 시퀀스자료형:
    반복할 코드
```

시퀀스에서 값을 하나씩 꺼내 변수에 넣고, 들여쓴 코드를 실행합니다.
값을 모두 꺼내면 반복이 끝납니다.

조건문과 마찬가지로 콜론과 들여쓰기가 필요합니다.

````python
print("시작")

for i in range(5): #0 1 2 3 4
    print(i)

print("끝")
````

````python
fruits = ["사과", "바나나", "포도"]
for fruit in fruits:
    print(fruit)
````

````python
for fruit in ["사과", "바나나", "포도"]:
    print(fruit)
````

````python
print("시작")

#for i in range(5): #0 1 2 3 4
for _ in range(5): #0 1 2 3 4
    print("hello")

print("끝")
````

````python
for fruit in ["사과", "바나나", "포도"]:
    print("hello")
````

### 실습 1

`range`와 `for`를 사용해 1부터 9까지 3단 구구단을 출력하세요.

출력 예시
```
3 x 1 = 3
3 x 2 = 6
...
3 x 9 = 27
```

````python
for i in range(1,10):
    print(f'3 X {i} = {3*i}')
````

````python
#반복 대상이 비어 있으면 본문은 한 번도 실행되지 않습니다.
for item in []:
    print("이 줄은 실행되지 않습니다")
print("반복 끝")
````

````python
total = 0                # 반복 전에 준비

for i in range(1, 11):
    total = total + i    # 반복하며 누적
    print(f"i={i}, total={total}")

print("합계:", total)
````

---
다음 점수의 평균을 구하는 for 반복문 코드를 작성하시오  
scores = [90, 85, 77, 92, 68]

---


````python
scores = [90, 85, 77, 92, 68]
total = 0
avg = 0
for score in scores:
    total = total + score
    #total += score

avg = total/len(scores)

print(f"{scores}의 평균은 {avg}입니다")
````

````python
for i in range(2, 10):
    for j in range(1, 10):
        print(f"{i} X {j} = {i*j}")
````
