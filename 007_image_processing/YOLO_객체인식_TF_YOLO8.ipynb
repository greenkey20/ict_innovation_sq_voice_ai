```python
# YOLOv8 객체 탐지 노트북

# 1. 필요 패키지 설치
# !pip install ultralytics

from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

# 2. 모델 로드 (YOLOv8n - 가장 가벼운 모델)
model = YOLO("yolov8n.pt")  # yolov8s.pt, yolov8m.pt 등도 가능

# 3. 이미지 로딩
img_path = "Yolo_Test_Image.jpg"  # 기존 이미지 경로 그대로 사용 TODO

# 4. 모델 추론
results = model(img_path)

# 5. 결과 시각화
results[0].show()  # bounding box가 포함된 이미지 자동 표시

# 6. 이미지 저장
results[0].save(filename="output_image_v8_강사님.jpg")

# 7. 탐지된 객체 출력
for box in results[0].boxes:
    cls = int(box.cls[0])
    conf = float(box.conf[0])
    print(f"Class: {model.names[cls]}, Confidence: {conf:.4f}")
```
