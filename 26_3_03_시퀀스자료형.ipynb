```python
text = "Python (파이썬)" # 문자열
numbers = [10, 20, 40] # 리스트
points = (3, 5) # 튜플

print(len(text), len(numbers), len(points))
# print(text[12]) # string index out of range
print(text[0], numbers[2], points[-2])
```

slicing은  +/- index 모두 사용 가능함

```python
word = "Python"
print(word[0:0])
print(word[0:1])
print(word[2:5])
print(word[:5])
print(word[2:])
print(word[::])
print(word[::])
print(word[-4::2])
```

```python
national_id = "000101-3456789"

birth_yymmdd = national_id[:6]
gender = national_id[7]

print(f"생년월일 = {birth_yymmdd}")
print(f"성별 = {gender}")

if gender == "1" or gender == "3":
    print("남자")
elif gender == "2" or gender == "4":
    print("여자")
else:
    print("입력값 오류")
```

```python
my_list = [4, 1, 2, 0, 17]
print(my_list[:-2:2])
print(my_list[2:5])
```

```python
my_list = [4, 1, 2, 0, 17] 
my_tuple = (4, 1, 2, 0, 17) # 값이 변경되면 안 되는 시퀀스 자료형
print(my_list)
print(my_tuple)

my_list[1] = 5
print(my_list)

my_tuple[1] = 5 # 'tuple' object does not support item assignment 에러 발생 <- 튜플은 읽기 전용 list, 값이 변경되지 않음
```

```python
my_range = range(10)
print(my_range)
print(list(my_range))

# my_range2 = range(-10.1, 20, 3) # 'float' object cannot be interpreted as an integer
my_range2 = range(-10, 20, 3)
print(my_range2)
print(list(my_range2))
```

---
1. 시퀀스 자료형: 모두 순서가 있다
   * 문자열: 파이썬에서 "" 사이에 들어가는 값의 자료형
   * 리스트: [] 안에 ,로 구분해서 원소를 쓰는 자료형
   * 튜플: () 안에 ,로 구분해서 원소를 나열하는 자료형
   * 레인지: range()라는 함수?를 사용해서 특정 범위 안에서 특정 간격의 정수를 리스트로 만들 수 있는 자료형
2. 시퀀스 자료형의 인덱스
   * [] 안에 :로 구분해서 자연수를 쓴다
   * 시작(0, 생략 가능) : 끝(-1) : 간격(1, 생략 가능)
3. 시퀀스 자료형 중 수정이 가능한 자료형: 리스트

> **강사님 답변**
> 
> 1. 시퀀스 자료형 4가지와 특징
> 
> list : 여러 데이터를 순서대로 저장하며, 저장된 값을 수정할 수 있습니다.
>
> tuple : 여러 데이터를 순서대로 저장하지만, 저장된 값을 수정할 수 없습니다.
>
> range : 일정한 규칙의 정수 범위를 만들어 반복문 등에 사용합니다.
>
> string : 여러 문자를 순서대로 저장하는 문자열 자료형이며, 문자 자체는 수정할 수 없습니다.
>
> 2. 시퀀스 자료형의 인덱스 인덱스는 시퀀스 자료형의 각 데이터에 붙는 위치 번호이며, 첫 번째 데이터의 인덱스는 0부터 시작합니다. 음수 인덱스를 사용하면 뒤에서부터 접근할 수 있으며, 마지막 데이터는 -1입니다.
>
> 3. 수정이 가능한 시퀀스 자료형 list만 수정 가능합니다. tuple, range, string은 생성된 후 값을 변경할 수 없습니다.

```python
test = "Python"
test[0] = "2"
print(test)
```
