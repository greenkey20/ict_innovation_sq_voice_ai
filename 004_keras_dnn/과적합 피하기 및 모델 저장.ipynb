```python
import pandas as pd
print(pd.__version__)

df = pd.read_csv('../dataset/sonar.csv', header=None)
print(df.info())
print(df.head())
```

```python
dataset = df.values
X = dataset[:, 0:-1].astype(float)
Y_obj = dataset[:, -1]

from sklearn.preprocessing import LabelEncoder

e = LabelEncoder()
e.fit(Y_obj)
Y = e.transform(Y_obj)

print(Y_obj)
print(Y)

import tensorflow as tf

Y_encoded = tf.keras.utils.to_categorical(Y)
print(Y_encoded)
```

```python
from sklearn.model_selection import train_test_split
import numpy as np

seed = 0
np.random.seed(seed)
tf.random.set_seed(seed)

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=seed)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential()
model.add(Dense(32, input_dim=60, activation='relu')) 
model.add(Dense(8, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X_train, Y_train, epochs=200, batch_size=5)

# print("\n Accuracy: %.4f" % (model.evaluate(X, Y)[1]))

# 학습 set으로 모델 검증
print("\n Train Accuracy: %.4f" % (model.evaluate(X_train, Y_train)[1]))

# 검증/test set으로 모델 검증
print("\n Validation/Test Accuracy: %.4f" % (model.evaluate(X_test, Y_test)[1]))
```

```python
model.save('../models/my_model.keras')
print("모델이 'my_model.keras' 파일로 저장되었습니다")
```
