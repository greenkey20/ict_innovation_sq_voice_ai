1. 딥러닝 모델 학습 결과는 실제 값과 예측 값의 차이(오차)를 최소화하는 선형 회귀식의 파라미터들이다. 이 산출물로 구성한 선형 회귀식에 실제 값을 넣어 해당 데이터에 대한 예측 값을 얻도록 활용한다.
2. 선형 회귀에서 '선형'은 독립변수와 종속변수의 관계를 직선으로 표현하고자 함을 의미한다.
3. 선형 회귀에서 오차는 실제 값과 예측 값의 차이로, 모델의 성능 지표로 역할한다.
4. 경사하강법은 손실함수의 값을 최소화하는 파라미터를 찾는 알고리즘이다. 경사하강법은 복잡한 손실함수에서 최소값의 위치를 직접 계산하기 어려우므로 반복적인 최적화를 통해 찾아가기 위해 사용한다.
5. 오차는 손실함수로 수치화되고, 경사하강법은 이 손실함수를 감소시키도록 모델의 파라미터를 조정하는 관계가 있고, 이를 통해 궁극적으로 달성하고자 하는 목표는 오차를 최소화하는 것이다.
