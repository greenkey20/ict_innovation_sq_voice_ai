## while 반복문

반복 횟수가 정해져 있지 않고 **조건이 참인 동안** 반복할 때 사용합니다.

```
while 조건:
    반복할 코드
```

조건이 거짓이 되는 순간 반복이 멈춥니다.
따라서 반복문 안에서 조건에 영향을 주는 값이 반드시 변해야 합니다.

````python
count = 1

while count <= 5:
    print(f"{count}번째")
    count = count + 1
    
print("종료")
````

### while의 세 가지 요소

1. 반복 전에 변수를 준비한다. (`count = 1`)
2. 조건을 검사한다. (`count <= 5`)
3. 반복 안에서 변수를 바꾼다. (`count += 1`)

세 번째를 빠뜨리면 조건이 영원히 참이어서 프로그램이 멈추지 않습니다.
주피터 노트북에서 무한 루프에 빠지면 상단의 중지(정사각형) 버튼으로 실행을 멈출 수 있습니다.

2. 사용자로부터 나이를 입력 받음

4. -1을 입력 하면 종료

### break

반복을 즉시 빠져나옵니다. `while True`와 함께 자주 사용합니다.

````python
while True: #무한 반복
    age = int(input("나이를 입력하세요(-1 입력 시 종료): "))

    if age == -1:
        print("프로그램을 종료합니다.")
        break

    if age >= 18:
        print("성년입니다.")
    else:
        print("미성년입니다.")
````
