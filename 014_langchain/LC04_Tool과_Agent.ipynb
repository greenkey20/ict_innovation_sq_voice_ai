```python
import os                         # 환경변수 읽기용
from dotenv import load_dotenv    # .env 파일을 읽어 환경변수로 등록하는 함수

load_dotenv()  # .env 파일에서 OPENAI_API_KEY를 읽어 옴

# 키가 없으면 여기서 바로 멈추고 안내 메시지 출력 (뒤에서 알기 어려운 에러가 나는 것을 방지)
assert os.getenv("OPENAI_API_KEY"), ".env 파일에 OPENAI_API_KEY가 없습니다."

# 실습 전체에서 사용할 모델 (gpt-4o-mini는 2027-01-14 종료 예정, 이후에는 여기만 바꾸면 됨)
MODEL = "gpt-4o-mini"
```

```python
from langchain.tools import tool   # 일반 함수를 LLM이 쓸 수 있는 도구로 바꿔 주는 데코레이터

# @tool 을 붙이면 함수 이름, 매개변수 타입, docstring이 LLM에게 도구 설명서로 전달됨
@tool
def add_numbers(a: int, b: int) -> int:
    """두 정수 a와 b를 더한 값을 반환합니다."""
    return a + b

@tool
def multiply_numbers(a: int, b: int) -> int:
    """두 정수 a와 b를 곱한 값을 반환합니다."""
    return a * b

# 도구는 LLM 없이도 직접 실행해 볼 수 있음 (매개변수는 딕셔너리로 전달)
print(add_numbers.invoke({"a": 3, "b": 5}))
print(multiply_numbers.invoke({"a": 4, "b": 6}))
print(add_numbers.name, "-", add_numbers.description)   # LLM이 보게 되는 도구 이름과 설명
```
