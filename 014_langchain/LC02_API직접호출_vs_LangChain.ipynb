```python
import os
from dotenv import load_dotenv

load_dotenv()  # .env 파일에서 OPENAI_API_KEY를 읽어 옴
assert os.getenv("OPENAI_API_KEY"), ".env 파일에 OPENAI_API_KEY가 없습니다."

# 실습 전체에서 사용할 모델 (gpt-4o-mini는 2027-01-14 종료 예정, 이후에는 여기만 바꾸면 됨)
MODEL = "gpt-4o-mini"
```

## example 1

```python
from openai import OpenAI

client = OpenAI()   # OPENAI_API_KEY 환경변수를 자동으로 사용

city = input("지역을 입력하세요 : ") or "부산"
days = input("여행 기간을 입력하세요 (예: 3일) : ") or "3일"

prompt = f"""
당신은 전문 여행 플래너입니다.
도시: {city}
여행 기간: {days}

{city}를 여행하는 {days} 일정표를 작성해 주세요.
각 일자별로 오전/오후 활동을 구체적으로 포함하세요.
"""

response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "당신은 여행 계획을 잘 짜는 어시스턴트입니다."},
        {"role": "user", "content": prompt},
    ],
    temperature=0.7, # 값이 작으면 매번 비슷한 vs 2에 가까우면 다양한 답변
)

print(f"사용자가 입력한 여행지 = {city}")
print(f"사용자가 입력한 여행 기간 = {days}")

print("\n=== 여행 일정표 (API 직접 호출) ===")
print(response.choices[0].message.content)
```

## example 2

```python
def suggest_brand_name_api(product: str) -> str:
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": "당신은 훌륭한 브랜딩 전문가입니다."},
            {"role": "user", "content": f"새로운 상품 '{product}'에 어울리는 브랜드 이름을 제안해 주세요."},
        ],
        temperature=0.7,
    )
    return response.choices[0].message.content.strip()


product_name = "알록달록 양말"

print(f"상품 = {product_name}")

print("\n=== API 직접 호출 ===")
print(f"상품명: {product_name}")
print(suggest_brand_name_api(product_name))
```
