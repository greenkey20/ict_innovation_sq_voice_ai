# LC01. LangChain 개념과 첫 번째 체인

**학습 목표**
- LLM을 앱에 넣을 때 필요한 흐름을 이해한다.
- LangChain의 핵심 구성 요소를 안다.
- `프롬프트 | 모델 | 출력 파서` 체인을 직접 만들어 실행한다.

## 1. 배경 지식

### 1-1. API만으로 앱을 만들 때의 불편함
- 프롬프트를 매번 문자열로 직접 조립해야 함
- 검색, 계산기 같은 외부 도구 연결이 번거로움
- 대화 기록 관리 코드를 직접 짜야 함

### 1-2. 그래서 LangChain
LLM 앱에 필요한 부품(프롬프트, 모델, 출력 처리, 기억, 도구)을 **조립하듯 연결**하게 해 주는 파이썬 프레임워크입니다.

## 2. LangChain 핵심 구성 요소

| 구성 요소 | 역할 | 비유 |
|---|---|---|
| PromptTemplate | 빈칸이 있는 프롬프트 틀 | 질문지 양식 |
| ChatModel | GPT 등 언어 모델 | 글 잘 쓰는 작가 |
| OutputParser | 모델 응답을 문자열/JSON 등으로 정리 | 결과 정리 담당 |
| Chain (`\|`) | 위 부품을 순서대로 연결 | 컨베이어 벨트 |
| Memory | 이전 대화 기록 | 대화 노트 |
| Tool | 계산기, 검색 등 외부 기능 | 도구 상자 |
| Agent | 상황에 맞는 도구를 스스로 골라 사용 | 비서 |

```
[사용자 입력] → PromptTemplate → ChatModel → OutputParser → [결과]
```

예전 자료의 `LLMChain`, `chain.run()`은 더 이상 쓰지 않습니다. 지금은 `prompt | llm` 형태로 연결하고 `invoke()`로 실행합니다.

# LangChain과 LangGraph, LLM 제공자 연동

## 1. LangChain과 LangGraph의 관계

**LangChain**은 LLM을 활용한 프로그램을 만들 때 프롬프트, 모델, 검색 도구 등을 연결해 주는 프레임워크입니다. 여러 작업을 순서대로 연결해 하나의 처리 과정을 만들 수 있습니다.

**LangGraph**는 여러 처리 단계를 그래프 형태로 구성하고, 각 단계의 상태와 실행 순서를 제어하는 프레임워크입니다. 조건에 따라 다른 단계로 이동하거나, 필요하면 앞 단계로 돌아가는 흐름을 만들 수 있습니다.

| 구분 | LangChain | LangGraph |
|---|---|---|
| 주요 역할 | 모델·프롬프트·도구 등 LLM 구성 요소 연결 | 노드와 상태를 이용해 복잡한 실행 흐름 제어 |
| 처리 방식 | 작업을 연결해 순차적으로 실행 | 노드와 간선으로 실행 경로를 구성 |
| 잘 맞는 경우 | 단순한 질의응답, 요약, 기본적인 RAG | 조건 분기, 반복, 오류 처리, 여러 단계의 에이전트 |

LangChain이 먼저 등장했고, **LangGraph는 복잡한 에이전트 흐름을 더 세밀하게 제어하기 위해 2024년 초 소개**됐습니다. 처음에는 LangChain 위에서 작동하고 LangChain 구성 요소와 호환되는 방식으로 설명됐습니다. 현재는 별도 패키지로 사용할 수 있으며, 두 프레임워크를 함께 쓰거나 필요한 기능만 선택해 사용할 수 있습니다.   

### 비유로 이해하기

- **LangChain**은 필요한 부품을 연결해 기능을 만드는 도구 상자에 가깝습니다.
- **LangGraph**는 그 부품들이 어떤 순서와 조건으로 움직일지 정하는 작업 흐름도에 가깝습니다.

두 도구는 서로 경쟁하는 관계라기보다 **다루는 부분이 다른 도구**입니다.

## 2. 주요 LLM 제공자와 연결하기

LangChain은 OpenAI, Google, Anthropic을 비롯한 여러 모델 제공자와 연동할 수 있습니다. 제공자별 연동 패키지를 설치하고 해당 서비스의 API 키를 준비해야 합니다.   

| 제공자 | LangChain 연동 예 |
|---|---|
| OpenAI GPT 모델 | `langchain-openai`의 `ChatOpenAI` |
| Google Gemini | `langchain-google-genai`의 `ChatGoogleGenerativeAI` |
| Anthropic Claude | `langchain-anthropic`의 `ChatAnthropic` |
| 그 밖의 제공자 | Mistral, Cohere, Ollama 등 다양한 연동 지원 |

연동 패키지는 제공자마다 다르지만, 패키지를 항상 해당 AI 회사가 직접 만들고 관리하는 것은 아닙니다. LangChain 팀이나 파트너가 관리하는 패키지도 있습니다.

### ChatGPT와 OpenAI API는 구분하기

LangChain이 연결하는 것은 **ChatGPT 웹사이트가 아니라 OpenAI가 제공하는 모델 API**입니다. 따라서 OpenAI API 키와 API 사용 권한이 필요합니다. Gemini와 Claude도 각 제공자의 API를 이용합니다.

또한 모델을 연결할 수 있다고 해서 모든 기능이 동일한 것은 아닙니다. 도구 호출, 이미지 입력, 구조화된 응답 등의 지원 여부와 사용 방법은 제공자와 모델에 따라 달라질 수 있습니다.

## 3. 이번 실습 코드와의 관계

이번 실습 코드에서는 LangGraph로 검색과 답변 생성의 순서를 구성했습니다.

```python
workflow = StateGraph(QAState)
workflow.add_node("web_search", web_search)
workflow.add_node("generate_answer", generate_answer)
```

하지만 답변 생성 단계에서는 LangChain 모델 객체가 아니라 OpenAI Python SDK를 직접 호출합니다.

```python
response = client.chat.completions.create(...)
```

따라서 현재 구조는 **LangGraph로 흐름을 관리하고, OpenAI SDK로 모델을 호출하는 방식**입니다. LangChain을 반드시 사용해야 LangGraph를 쓸 수 있는 것은 아닙니다.

LangChain의 모델 연동 객체를 사용한다면 다음처럼 구성할 수 있습니다.

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="사용 가능한 모델 이름")
response = model.invoke("질문을 입력합니다")
```

이 경우에도 실제 답변은 OpenAI 모델이 생성합니다. LangChain은 모델 호출 방법을 일정한 형태로 다룰 수 있도록 연결해 주는 역할을 합니다.

## 핵심 정리

1. **LangChain**은 LLM, 프롬프트, 검색기, 도구 등을 연결하는 데 도움을 줍니다.
2. **LangGraph**는 노드와 상태를 이용해 실행 순서, 조건 분기, 반복을 관리합니다.
3. LangGraph는 LangChain보다 나중에 소개됐으며, 지금은 두 도구를 따로 또는 함께 사용할 수 있습니다.
4. LangChain은 GPT, Gemini, Claude 등 여러 제공자와 연동할 수 있지만, 각각의 API 키와 연동 패키지가 필요합니다.
5. 이번 실습은 **LangGraph와 OpenAI SDK를 직접 연결한 구조**이며 LangChain은 사용하지 않습니다.

# 음성 모델과 LangChain / LangGraph의 관계

> 음성 모델의 **개념**(원리)을 이해하고, 실제 서비스에서는 **이미 만들어진 음성 모델을 부품으로 활용**한다.
> 이 부품들을 엮어 하나의 서비스로 만드는 도구가 **LangChain / LangGraph**다.

---

## 1. 전체 그림

| 단계 | 무엇을 배우나 | 질문 |
|---|---|---|
| 1. LLM 원리 | 토큰, 임베딩, 어텐션, 다음 토큰 예측 | 언어 모델은 어떻게 생성하는가 |
| 2. 음성 모델 개념 | 멜 스펙트로그램, 코덱 토큰, 보코더 | 소리를 어떻게 토큰으로 다루는가 |
| 3. 음성 모델 활용 | Whisper, CosyVoice 등 사전학습 모델 사용 | 만들어진 모델을 어떻게 가져다 쓰는가 |
| 4. 서비스 구성 | LangChain / LangGraph | 여러 모델을 어떻게 하나의 흐름으로 엮는가 |

---

## 2. 음성 모델의 개념: LLM 원리의 확장

최근 음성 모델은 **소리를 토큰으로 바꾼 뒤, 언어 모델처럼 다음 토큰을 예측**하는 구조로 수렴하고 있다.

| 모델 | 역할 | 구조 핵심 |
|---|---|---|
| Whisper | STT (음성 → 텍스트) | 멜 스펙트로그램 → 트랜스포머 인코더 → 디코더가 텍스트 토큰 생성 |
| VALL-E | TTS (텍스트 → 음성) | 오디오 코덱으로 음성을 이산 토큰화 → 다음 음성 토큰 예측 ("소리를 다루는 GPT") |
| CosyVoice | TTS (텍스트 → 음성) | LLM이 음성 토큰 생성 → flow matching으로 멜 복원 → 보코더로 파형 생성 |

### LLM과 공통인 개념
- 토큰화, 임베딩
- 어텐션, 트랜스포머
- 자기회귀 생성 (다음 토큰 예측)
- 샘플링 (temperature 등)

### 음성에만 있는 개념
- 멜 스펙트로그램: 소리를 사람의 청각에 맞춘 주파수 그림으로 표현
- 오디오 코덱과 양자화(RVQ): 연속적인 소리를 이산 토큰으로 변환
- 보코더: 스펙트로그램이나 토큰을 실제 파형으로 복원
- 화자 임베딩: 목소리의 특징을 벡터로 표현

> 정리: **텍스트 토큰 자리에 코덱 토큰이 들어간 것**이라고 이해하면, 음성 모델은 LLM 원리의 응용으로 자연스럽게 읽힌다.

---

## 3. 음성 모델의 활용: LangChain / LangGraph의 역할

LangChain / LangGraph는 음성 모델의 **원리를 설명하는 도구가 아니다.**
이미 만들어진 음성 모델(STT, TTS)과 LLM을 **연결해 서비스로 만드는 도구**다.

### 음성 에이전트 파이프라인

```
음성 입력
   ↓
[STT 노드]   Whisper 등으로 음성 → 텍스트
   ↓
[LLM 노드]   질문 이해, 판단
   ↓ (필요 시 분기)
[검색/도구 노드]   매뉴얼 검색(RAG), 설비 조회 등
   ↓
[TTS 노드]   CosyVoice 등으로 텍스트 → 음성
   ↓
음성 출력
```

| 구성 요소 | 담당 | 비고 |
|---|---|---|
| STT | 음성 모델 | 사전학습 모델 활용 |
| 판단, 대화 | LLM | |
| 문서 검색 | RAG (LangChain) | 벡터DB, retriever |
| 도구 호출 | Tool / Agent (LangChain) | |
| 전체 흐름, 분기, 상태 | LangGraph | 노드와 엣지로 구성 |
| TTS | 음성 모델 | 사전학습 모델 활용 |

### LangGraph 구조 예시 (뼈대)

```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

class VoiceState(TypedDict):
    audio_in: bytes     # 입력 음성
    text: str           # STT 결과
    answer: str         # LLM 답변
    audio_out: bytes    # TTS 결과

def stt_node(state: VoiceState):
    text = ...          # 사용할 STT 모델 호출 (예: Whisper)
    return {"text": text}

def llm_node(state: VoiceState):
    answer = ...        # LLM 또는 LangGraph 상담 Agent 호출
    return {"answer": answer}

def tts_node(state: VoiceState):
    audio = ...         # 사용할 TTS 모델 호출 (예: CosyVoice)
    return {"audio_out": audio}

graph = StateGraph(VoiceState)
graph.add_node("stt", stt_node)
graph.add_node("llm", llm_node)
graph.add_node("tts", tts_node)
graph.add_edge(START, "stt")
graph.add_edge("stt", "llm")
graph.add_edge("llm", "tts")
graph.add_edge("tts", END)
app = graph.compile()
```

`llm_node` 자리에 LG04의 상담 Agent를 그대로 넣으면 **음성으로 묻고 답하는 매뉴얼 상담봇**이 된다.

### 활용 예
- 음성 상담봇 (제품 매뉴얼 기반 RAG + 음성 입출력)
- 음성 비서 (일정, 검색, 도구 호출)
- 제조 현장 음성 작업 지시와 설비 상태 음성 조회

## 3. 첫 번째 체인: 프롬프트 | 모델

````python
# 1. 필요한 도구 불러오기
from langchain_openai import ChatOpenAI           # OpenAI 채팅 모델을 LangChain에서 쓰게 해 주는 클래스
from langchain_core.prompts import PromptTemplate  # 빈칸({변수})이 있는 프롬프트 틀을 만드는 클래스

# 2. 모델 준비
#    model       : 사용할 모델 이름 (앞 셀에서 MODEL = "gpt-4o-mini"로 정함)
#    temperature : 답변의 다양성 (0이면 매번 비슷한 답, 높을수록 다양한 답, 보통 0~1 사이)
llm = ChatOpenAI(model=MODEL, temperature=0.7)

# 3. 프롬프트 틀 만들기
#    {place} 자리는 실행할 때 원하는 값으로 채워짐 (서울, 부산 등)
prompt = PromptTemplate.from_template("여행지 {place}에 대한 3일 일정을 짜 줘")

# 4. 체인 만들기
#    | 기호로 부품을 연결: 프롬프트가 완성한 문장을 모델에 전달
chain = prompt | llm

# 5. 체인 실행
#    딕셔너리의 키 이름("place")이 프롬프트의 {place}와 같아야 함
#    실행 순서: {place}에 "서울"을 채움 → 완성된 문장을 모델에 보냄 → 답변 받음
result = chain.invoke({"place": "서울"})

# 6. 결과 확인
print(type(result).__name__)   # AIMessage: 답변 텍스트와 토큰 사용량 등 부가 정보를 함께 담은 객체
print(result.content)          # .content 로 실제 답변 텍스트만 꺼냄
````

## 4. 출력 파서 붙이기
모델은 `AIMessage` 객체를 돌려줍니다. 끝에 `StrOutputParser()`를 붙이면 바로 문자열을 받습니다.

````python
# 1. 출력 파서 불러오기
#    StrOutputParser: 모델이 돌려준 AIMessage 객체에서 답변 텍스트(문자열)만 꺼내 주는 부품
from langchain_core.output_parsers import StrOutputParser

# 2. 체인 만들기 (부품 3개 연결)
#    prompt           : {place}를 채워 완성된 질문 문장을 만듦
#    llm              : 질문을 받아 AIMessage 형태로 답변
#    StrOutputParser(): AIMessage에서 텍스트만 꺼냄 (뒤에 괄호 () 를 붙여 객체로 만들어야 함)
chain = prompt | llm | StrOutputParser()

# 3. 체인 실행
#    이번에는 결과가 AIMessage가 아니라 바로 문자열(str)로 나옴
text = chain.invoke({"place": "부산"})

# 4. 결과 출력
#    .content 없이 바로 출력 가능
print(text)
````

````python
# 확인
assert isinstance(text, str) and len(text) > 0
print("LC01 확인 완료")
````

````python
# 참고 : 위 코드를 claude API 로 사용하는 방법 - API 가 없어서 실행은 안됨
import os
from dotenv import load_dotenv

load_dotenv(override=True)  # .env 값으로 환경변수를 덮어씀
assert os.getenv("ANTHROPIC_API_KEY"), ".env 파일에 ANTHROPIC_API_KEY가 없습니다."

# 실습에서 사용할 Claude 모델 (Haiku: 빠르고 저렴, 실습용으로 적합)
MODEL = "claude-haiku-5-5"

# 1. 필요한 도구 불러오기
from langchain_anthropic import ChatAnthropic       # Claude 모델을 LangChain에서 쓰게 해 주는 클래스
from langchain_core.prompts import PromptTemplate    # 빈칸({변수})이 있는 프롬프트 틀

# 2. 모델 준비
#    max_tokens : 답변 최대 길이 (Claude API는 이 값을 반드시 함께 보냄)
llm = ChatAnthropic(model=MODEL, max_tokens=2048)

# 3. 프롬프트 틀 만들기 (OpenAI 버전과 동일)
prompt = PromptTemplate.from_template("여행지 {place}에 대한 3일 일정을 짜 줘")

# 4. 체인 만들기 (OpenAI 버전과 동일)
chain = prompt | llm

# 5. 실행
result = chain.invoke({"place": "서울"})
print(type(result).__name__)   # AIMessage: OpenAI와 같은 형태로 반환됨
print(result.content)          # 실제 답변 텍스트
````

````python
# 출력 파서 붙이기 - API 가 없어서 실행은 안됨
from langchain_core.output_parsers import StrOutputParser

# 프롬프트 → Claude → 문자열
chain = prompt | llm | StrOutputParser()

text = chain.invoke({"place": "부산"})
print(text)
````
