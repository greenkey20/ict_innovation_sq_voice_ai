- 칸 번호는 tokenizer.word_index의 번호와 같아요. 예상되는 번호는 the:1, cat:2, sat:3, on:4, mat:5, dog:6, ate:7, my:8, homework:9예요.
- 0번 칸은 항상 0이에요(번호가 1부터 시작해서 비어 있는 칸).

그래서 첫 문장(The cat sat on the mat.)은 1~5번 칸이 1이고, 둘째 문장(The dog ate my homework.)은 1, 6, 7, 8, 9번 칸이 1이에요. 둘 다 the(1번)가 있어서 첫 칸이 1이고, 둘째 칸이 1이냐 0이냐는 cat이 있느냐(2번)로

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Flatten, Dense
import warnings

warnings.filterwarnings('ignore')

# 10개 dummy data(가짜, 작은 사이즈의) 리스트, 클래스(긍정 or 부정 등?)가 없는 문장, 굳이 분류하자면 왕 1 vs 여왕 0으로 할 수 있겠지만, 10개로는 어차피 훈련 안 됨
sentences = [
    "왕은 나라를 다스리는 남자입니다.",
    "여왕은 나라를 다스리는 여자입니다.",
    "남자가 왕이 되었습니다.",
    "여자가 여왕이 되었습니다.",
    "왕과 여왕은 함께 궁전에서 살고 있습니다.",
    "남자와 여자가 파티에 참석했습니다.",
    "왕은 충성스러운 신하들을 거느립니다.",
    "여왕은 아름다운 드레스를 입고 있었습니다.",
    "남자가 여왕의 조언자가 되었습니다.",
    "여자가 왕의 자리를 노리고 있습니다."
]

tokenizer = Tokenizer(
    # 조사 붙은 단어까지 구분되도록 최소 필터만 적용
    filters='!"#$%&()*+,-./:;<=>?@[\\]^_`{|}~\t\n'
)
```

```python
tokenizer.fit_on_texts(sentences)

print("-- 참고 --")
print(tokenizer.word_index)
print(tokenizer.word_counts)
print(tokenizer.document_count)
print(tokenizer.word_docs)
print("----------")

sequences = tokenizer.texts_to_sequences(sentences) # 각 문장의 토큰들이 숫자로 바뀜?

word_index = tokenizer.word_index
print("단어 목록/사전 = word_index = ", word_index)

padded = pad_sequences(sequences, padding='post')
print("패딩된 시퀀스 = \n", padded)

vocab_size = len(word_index) + 1
embedding_dim = 8 # 간단한 실습은 2, 4, 8개 정도면 됨 vs Naver 영화 리뷰 감정 분류 5천개 데이터 2세트는 32, 64개 vs 언어 모델? 128, 256개 vs 조금 더 큰 모델은 1024개 vs gpt 같은 건 수천/몇만 차원..
max_len = padded.shape[1]
```

```python
model = Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=embedding_dim)) # input_length=max_len, embedding 벡터 layer
model.add(Flatten()) # 분류기로 변환?
model.add(Dense(1, activation='sigmoid')) # 출력층 맞나? 여기는 왜 sigmoid 활성화 함수?

model.compile(optimizer='adam', loss='binary_crossentropy') # metrics=['accuracy'], 2진 분류, 이 예시에서는 의미가 없기 때문에 평가지표 accuracy 안 썼지만 감정 분류 등에는 유용할 수 있음
model.summary()

model.fit(padded, np.ones((len(sentences), 1)), epochs=10, verbose=2) # x = padded, y = dummy 1행렬
model.summary()

embedding_matrix = model.layers[0].get_weights()[0]
```

## imdb 학습 기본 코드

```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras import preprocessing

# 특성으로 사용할 단어의 수
max_features = 10000

# 사용할 텍스트의 길이(가장 빈번한 max_features 개의 단어만 사용합니다)
maxlen = 20

# 정수 리스트로 데이터를 로드합니다.
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)

# 리스트를 (samples, maxlen) 크기의 2D 정수 텐서로 변환합니다.
x_train = preprocessing.sequence.pad_sequences(x_train, maxlen=maxlen)
x_test = preprocessing.sequence.pad_sequences(x_test, maxlen=maxlen)

print(x_train.shape, y_train.shape)
print(x_test.shape, y_test.shape)
print(x_train[0])
print(y_train[0])
```

```python
# IMDB 데이터셋의 단어 인덱스 로드
word_index = imdb.get_word_index()

# 인덱스를 단어로 변환하는 딕셔너리 생성
index_to_word = {index + 3: word for word, index in word_index.items()}
index_to_word[0] = "<PAD>" # padding
index_to_word[1] = "<START>" # 문장 시작
index_to_word[2] = "<UNK>" # unknown
index_to_word[3] = "<UNUSED>" # 미사용

# 숫자 시퀀스를 실제 단어 시퀀스로 변환하는 함수
def decode_review(encoded_review):
    return ' '.join([index_to_word.get(i, '?') for i in encoded_review])

# 첫 번째 훈련 데이터 시퀀스를 실제 단어로 변환
decoded_review = decode_review(x_train[0])
print(decoded_review)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense, Embedding

print(x_train.shape)

model = Sequential()
# 나중에 임베딩된 입력을 Flatten 층에서 펼치기 위해 Embedding 층에 input_length를 지정합니다.
model.add(Embedding(10000, 8, input_length=maxlen))
# 임베딩 텐서를 (samples, maxlen * 8) 크기의 텐서로 펼칩니다.
model.add(Flatten())
# 분류기를 추가합니다.
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['acc']) # rmsprop = adam 바로 직전의 optimizer로서 사용해도 무방
model.summary() # 모델 한 번이라도 돌려야 의미있는 값들 볼 수 있음

history = model.fit(x_train, y_train,
                    epochs=10,
                    batch_size=32,
                    validation_split=0.2)

model.summary()
```

```python
# 임베딩 행렬 추출
embedding_layer = model.layers[0] # 0번째 층이 Embedding 층
embedding_matrix = embedding_layer.get_weights()[0]  # [0] = 단어 embedding

print("임베딩 행렬 모양:", embedding_matrix.shape)  # shape = (10000, 8) = (단어 수, 차원)

# 정수 인덱스 → 단어 매핑 준비
word_index = imdb.get_word_index()  # 단어 → 인덱스
index_to_word = {i + 3: w for w, i in word_index.items()}

index_to_word[0] = "<PAD>"
index_to_word[1] = "<START>"
index_to_word[2] = "<UNK>"
index_to_word[3] = "<UNUSED>"

# 보고 싶은 단어의 임베딩 확인
def show_vector(word):
    idx = word_index.get(word)

    if idx is None or idx + 3 >= embedding_matrix.shape[0]:
        print("단어가 어휘 사전에 없습니다.")
        return

    vec = embedding_matrix[idx + 3]  # +3 오프셋
    print(f"'{word}' 임베딩:", vec)

show_vector("good")
show_vector("bad")
show_vector("movie")
show_vector("bomb")
```

```python
import re
from tensorflow.keras.datasets import imdb
from tensorflow.keras import preprocessing

# IMDB 단어 인덱스 로드
word_index = imdb.get_word_index()

# 새로운 리뷰
new_reviews = [
    "It was a boring movie. I do not recommend it.",
    "The movie was extremely disappointing.",
    "I really loved this movie. It was wonderful and excellent.",
    "This was an amazing and enjoyable movie. I highly recommend it."
]

# 새로운 리뷰를 정수 인덱스로 변환
new_sequences = []

for review in new_reviews:
    # 영문과 작은따옴표를 제외한 문장부호 제거
    cleaned_review = re.sub(r"[^a-zA-Z']", " ", review.lower())
    words = cleaned_review.split()

    # IMDB 입력 형식에 맞추어 시작 토큰 추가
    sequence = [1]  # <START>

    for word in words:
        index = word_index.get(word)

        if index is not None and index + 3 < max_features:
            sequence.append(index + 3)
        else:
            sequence.append(2)  # <UNK>

    new_sequences.append(sequence)

# 학습 때와 같은 길이로 패딩
padded_new_sequences = preprocessing.sequence.pad_sequences(
    new_sequences,
    maxlen=maxlen
)

# 변환된 문장 확인
for sequence in padded_new_sequences:
    print(decode_review(sequence))
```

```python
# 예측
predictions = model.predict(padded_new_sequences)
```

```python
# 결과 출력
for review, prediction in zip(new_reviews, predictions):
    probability = prediction[0]
    result = "긍정" if probability > 0.5 else "부정"

    print(f"리뷰: {review}")
    print(f"예측: {result} (긍정 확률: {probability:.4f})")
    print()
```
