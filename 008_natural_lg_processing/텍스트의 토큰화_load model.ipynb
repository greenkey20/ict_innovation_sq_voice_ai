## 영화 긍/부정 리뷰 분류 모델과 토크나이저를 load하고 예측

새로운 노트에서 작성

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import load_model
import pickle

# 토크나이저 로드
with open('tokenizer.pickle', 'rb') as handle:
    token = pickle.load(handle)

# 모델 로드
model = load_model('../models/sentiment_model.keras')
```

```python
# 새로운 문장을 사용한 검증
new_reviews = [
    "너무 재밌네요",
    "나는 별로예요",
    "잘 만든 영화입니다",
    "지루하네요 재미없어요"
]

new_sequences = token.texts_to_sequences(new_reviews)
print(new_sequences)
```

```python
padded_new_sequences = pad_sequences(new_sequences, 4)
print(padded_new_sequences)
```

```python
# 예측
predictions = model.predict(padded_new_sequences)
print(predictions)
```

```python
# 예측 결과 출력
for review, prediction in zip(new_reviews, predictions):
    print(
        f"리뷰: {review} -> "
        f"예측: {'긍정' if prediction[0] > 0.5 else '부정'} "
        f"(확률: {prediction[0]:.4f})"
    )
```
